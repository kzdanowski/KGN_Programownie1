```python
macierz = [
    [1, 2, 3],
    [4, 5, 6]
]

splaszczona = []

for wiersz in macierz:
    for element in wiersz:
        splaszczona.append(element)

splaszczona
```

```python
macierz = [
    [1, 2, 3],
    [4, 5, 6],
    [7, 8, 9]
]

# Odpowiednik: for wiersz in macierz: for element in wiersz: ...
# splaszczona = [element for wiersz in macierz for element in wiersz]
splaszczona = [x for y in macierz for x in y]

splaszczona
```
