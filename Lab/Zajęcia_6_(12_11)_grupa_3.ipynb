## Definicja silni
n! = n × (n−1)!, dla 𝑛 > 1 n>1, a 0!= 1

```python
def silnia_rec(n):
    print(f"Wyołanie silnia_rec({n})")
    if n == 0:
        print(f"Koniec wywołania silnia({n}), zwracana wartość: {1}")
        return 1
    x =  n*silnia_rec(n-1)
    print(f"Koniec wywołania silnia({n}), zwracana wartość: {x}")
    return x
```

```python
silnia_rec(5)
```

## Rekurencyjna funkcja sumy

Wykorzystaj schemat: head and tail: head = pierwszy element listy, tail: lista bez pierwszego elementu.

```python
def sum_rec(l: list) -> int:
  if l == []:
    return 0
  return l[0] + sum_rec(l[1:])


sum_rec([1,2,3,4,5])
```

## Ciąg Fibonacciego

Każdy wyraz to suma dwóch poprzednich:

𝐹(𝑛)=𝐹(𝑛−1)+𝐹(𝑛−2)F(n)=F(n−1)+F(n−2), z 𝐹(0)=0, F(1)=1

Fibonacci: 1,1,2,3,5,8,13,21...

```python
def fib(i: int) -> int:
  # if i in [1,2]:
  #   return 1

  if i <= 1:
    return i

  # if i == 1 or i == 2:
  #   return 1

  return fib(i-1) + fib(i-2)

fib(5)
```

```python
# optymalizacja rekurencji za pomocą lru_cache

from functools import lru_cache

@lru_cache(None)
def fib(i: int) -> int:
  print(f"wywołanie fib {i}")
  if i <= 1:
    print(f"koniec wywołania fib {i}, zwraca 1")
    return i
  x = fib(i-1) + fib(i-2)
  print(f"koniec fib {i}, zwraca wartość {x}")
  return x


fib(5)
```

```python
# memoizacja - optymalizacja rekurencji z pomocą wbudowanego słownika zapamiętującego obliczone wartości

def fib_dict(i:int, d: dict = {0:0, 1:1}) -> int:
  if i not in d:
    print(f"stan słownika przy {i}, to {d}")
    d[i] = fib_dict(i - 1) + fib_dict(i -2)
    print(f"stan słownika przy {i}, to {d}")
  return d[i]

fib_dict(5)
```

```python
# memoizacja za pomocą słownika, kóry jest inicjowany od nowa przy każdym wywołaniu funkcji (nie przechwuje w pamięci niepotrzebnych wartości )

def fib_dict_none(i:int, d: dict = None) -> int:

  if d == None:
    d = {0:0, 1:1}


  if i not in d:
    print(f"stan słownika przy {i}, to {d}")
    d[i] = fib_dict_none(i - 1) + fib_dict_none(i -2)
    print(f"stan słownika przy {i}, to {d}")
  return d[i]

fib_dict_none(5)
```

LRU (Least Recently Used) zapamiętuje wyniki funkcji.

Funkcja nie musi liczyć tego samego wiele razy

Wynik zostaje w pamięci

Optymalizuje czas obliczeń.

```python
# Będziemy używać dekoratora funkcji, jest to funkcja, która modyfikuje inne funkcje

def dekorator(f):
    def wrapper():
        print("Zaczynam…")
        wynik = f()
        print("Koniec.")
        return wynik
    return wrapper

@dekorator
def hello():
    print("Cześć!")

hello()
```

```python
import time
from functools import lru_cache

# --- Funkcja rekurencyjna bez cache ---
def fib_no_cache(n):
    if n < 2:
        return n
    return fib_no_cache(n-1) + fib_no_cache(n-2)

# --- Funkcja rekurencyjna z cache ---
@lru_cache(maxsize=None)
def fib_with_cache(n):
    if n < 2:
        return n
    return fib_with_cache(n-1) + fib_with_cache(n-2)

# ----Moja funkja ze słownikiem

def fib_dict(i:int, d: dict = {0:0, 1:1}) -> int:
  if i not in d:
    d[i] = fib_dict(i - 1) + fib_dict(i -2)
  return d[i]

# --- Pomiar czasu ---
def measure_time(func, *args):
    start = time.perf_counter()
    result = func(*args)
    end = time.perf_counter()
    print(f"{func.__name__}({args[0]}) = {result}")
    print(f"Czas wykonania: {end - start:.6f} sekundy\n")



measure_time(fib_no_cache, 35)
measure_time(fib_with_cache, 35)
measure_time(fib_dict, 35)
```
